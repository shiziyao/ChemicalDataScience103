```python
inventory= pd.read_csv("inventory_solids.csv")
inventory
```

```python
inventory.columns=['name','sum_formula','amount(g)']
inventory.columns
```

```python
inventory.loc[len(inventory)]=['Sodium Chloride','NaCl',100]

inventory
```

```python
inventory.drop([5,6],inplace=True)
inventory
```

```python
inventory2= pd.read_csv("inventory_solids.csv")
inventory2.columns=['name','sum_formula','amount(g)']
```

```python
inventory3= pd.concat([inventory, inventory2], axis=0, ignore_index=True)
inventory3
```

```python
inventory3.to_csv("complete_inventory.csv",index=False)
```

```python
inventory_grouped=inventory3.groupby(['name'])[['amount(g)']].sum()
inventory_grouped
```

```python
amount_mean_formula=inventory3.groupby(['sum_formula'])[['amount(g)']].mean()
amount_mean_formula
```
