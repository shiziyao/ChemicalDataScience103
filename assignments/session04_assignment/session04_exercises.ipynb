## Key Concepts Review

### Binary Numbers and Float Precision

Computers store numbers in binary (base 2), which can cause precision issues with decimal numbers:

```python
# Float precision issues
print(0.1 + 0.2)  # Output: 0.30000000000000004 (not exactly 0.3!)
print(0.1 + 0.2 == 0.3)  # Output: False

# Better way to compare floats
def close_enough(a, b, tolerance=1e-9):
    return abs(a - b) < tolerance

print(close_enough(0.1 + 0.2, 0.3))  # Output: True
```

### Objects and Types
Everything in Python (except keywords) is an object with a type:

```python
# Scalar types (single values)
concentration = 0.5      # float

# Check types

```

````python
# PSEUDO CODE
#the numbers can be of three types after removing preceeding 0's (000024.567): 
#CASE A: 0.12345 or 0.0012345  (0's before any non-zero digit starts and 0 before decimal)
#CASE B: 1.2345 or 1.02345 (where (len before decimal) < precision, so precision will cross decimal)
#CASE C: all numbers are non-zero or zero is in between, no preceeding 0's, and count  before'.'>precision

# Step 1: split the values into 2 parts at '.' --> values1, values2
# Step 2: remove all the preceeding 0's in values1, so that the preceeding 0's arent 
# counted in significant digit count
# Step 3: CASE A: if there are preceeding 0's, like a 0 before '.' and 0's after '.'
#                 so significant digits will start from the first non-0 digit
#                 if values1==0 and va;ues2[0]==0, then the count of 0's == [abs(log10(value))] 
#                 so the first non=0 digit will come after 0+'.'+abs(log10(value)) till precision
#
# Step 4: CASE B: if the count of digits before '.' < precision, it means the total no. of significant
#                  digits of the number crosses '.', so the number will be till precision 
# 
# Step 5: CASE C: if the count of digits before'.'> precision, it means the precision will end before '.',
#                 so replace all value[precision:] with 0   
````

````python
def round_to_significant_figures(value: float, precision: int) -> float:

    value=str(value)

    values1,values2= value.split(".") 
    values1=values1.lstrip('0')
    
    if values1=='' or int(values1)==0 and values2[1]=='0':     # for values like 0.0000025 and 0.1234
        #values2.lstrip('0')                                   # with 0._ _ _
        zeros_after_decimal=abs(int(math.log10(float(value))))
        value_new=''  
        for k in range(precision+1+1+zeros_after_decimal):     # +1 for decimal and +1 for range
            value_new=value_new+value[k]                                     

    elif len(values1)<precision: #and values2[0] !=0:          # for values like 1.234000 or 1.20345
        value_new=""                                           # before decimal digits len < precision
        for i in range(precision+1):
            value_new=value_new+value[i]

    else:                                                      # for values like 12345.6789 (all numerical)
            value_new=value.replace(value[(precision):],"0")   # or 123456.7891234 
              
    return float(value_new)                 
````

## Exercise 3: Reaction Simulator Class

**Objective:** Create a reaction simulator that models the concentration of the reagents and products as a reaction progresses. 

**Problem:** Build a ChemicalReaction class that simulates simple chemical reactions, tracks reactant consumption, and handles errors when invalid data is provided.

**Your Task:** Complete the class below:

````python
# PSEUDO CODE
# 1. define a class ChemicalReaction with 5 methods (functions with a particular use)
# 2. initialize the reactants, products, and reaction rate which will be input and then make 
#    a reaction vessel and add all the reactants and products to it
# 3. Check if the new input element is present in the reaction vessel or not else raise ValueError
# 4.  find the LR using the ratio --> min(all amount/coeff)
# 5. if we find the LR, and we have the max extent of reaction by reaction_rate x time (LR)
# 6. final amount of reactant in the vessel = initial amount of reactant - coeff x max extent of LR
# 7. final amount of product in vessel = initial amount of product + coeff x max extent of LR
# 8. return the reaction status finally, and if the amount matches with question
#    it will return silent True
````

````python
import math
from typing import Dict
from collections import defaultdict
from itertools import chain

class ChemicalReaction:
    def __init__(self, reactants: Dict[str, int], products: Dict[str, int], reaction_rate: int = 1.0):
        """
        Initialize a chemical reaction
        reactants: dict like {"H2": 2, "O2": 1}
        products: dict like {"H2O": 2}
        the terms reagent and reactant are used here interchangeably but reactant is strictly more correct
        The integer values for reactants and products indicate the reaction stoichiometry
        reaction_rate: float (how fast the reaction proceeds, in one reaction given by reactants and products per second)
        """

        # To be initialised
        self.reactants = reactants
        self.products = products
        self.reaction_rate = reaction_rate


        # Add all the reactants and products to the reaction vessel
        self.reaction_vessel_content: Dict[str, float] = { compound: 0.0 for compound in {**reactants, **products}}

    # Add a method to check if the input is present wither in reactant or product list otherwise raise error    
    def add_reactant_amount(self, compound: str, amount: float):
        """Add amount of reactant to the reaction vessel. What happens if the compound added is not part of the reagents or products?"""
        if compound not in self.reactants and compound not in self.products:
            raise ValueError(f"'{compound}' not present in either reactant or product !!!")
        
        self.reaction_vessel_content[compound]= self.reaction_vessel_content.get(compound,0.0)+amount  


    # Add a method to find LR
    def find_limiting_reagent(self):
        ratios={}
        for compound, coeff in self.reactants.items():
            available_in_vessel= self.reaction_vessel_content.get(compound,0.0)
            ratios[compound] = available_in_vessel/coeff

        limiting_reagent = min(ratios, key=ratios.get) 
        return limiting_reagent, ratios[limiting_reagent]   

    # Add a method to finally find the extent of reaction with LR
    def run_reaction(self, time_seconds: float = 1.0):
        """Run reaction for given time and update reactant amounts. What happens if not enough of one compound is left?"""
        self.extent_of_reaction = self.reaction_rate * time_seconds
  
        limiting_compound, limiting_ratio = self.find_limiting_reagent()
        self.extent_of_reaction = min(self.extent_of_reaction, limiting_ratio)

        for compound, coeff in self.reactants.items():
            self.reaction_vessel_content[compound] = self.reaction_vessel_content.get(compound,0.0)-coeff * self.extent_of_reaction

        for compound, coeff in self.products.items():
            self.reaction_vessel_content[compound]= (
                self.reaction_vessel_content.get(compound, 0.0) + coeff * self.extent_of_reaction
            )
        return self.extent_of_reaction

    def get_reaction_status(self):
        return self.reaction_vessel_content.copy()
````

````python
## Test case 1: General function of the simulator

# Add reactants
water_reaction.add_reactant_amount("H2", 10.0)
water_reaction.add_reactant_amount("O2", 3.0)

# Check initial state of reaction vessel
reaction_status = water_reaction.get_reaction_status() 
print(reaction_status)
assert reaction_status['H2'] == 10.0
assert reaction_status['O2'] == 3.0
assert reaction_status['H2O'] == 0

# Run reaction
water_reaction.run_reaction(2.0)  # Run for 2 timepoints

# Check content of reaction vessel again
reaction_status = water_reaction.get_reaction_status() 
print(reaction_status)
assert reaction_status['H2'] == 8.0
assert reaction_status['O2'] == 2.0
assert reaction_status['H2O'] == 2.0
````
